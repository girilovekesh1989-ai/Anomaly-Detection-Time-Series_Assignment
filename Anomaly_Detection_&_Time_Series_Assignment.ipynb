```python
'''
Question 1: What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.
'''
'''


[Anomaly detection – Harini Jayaraman](https://www.lancaster.ac.uk/stor-i-student-sites/harini-jayaraman/anomaly-detection/)
[What Is Anomaly Detection? Methods, Use Cases, Tools](https://coralogix.com/guides/anomaly-detection/)
[Data Anomalies: Types, Detection, and Business Impact](https://www.anomalo.com/blog/data-anomaly-what-is-it-common-types-and-how-to-identify-them/)
[Anomaly Detection Algorithms: Types & Techniques| Fidelis ...](https://fidelissecurity.com/threatgeek/network-security/anomaly-detection-algorithms/)
[Anomaly detection - Things Solver](https://thingsolver.com/blog/anomaly-detection/)

Anomaly detection is the identification of rare items, events, or observations that deviate significantly from expected or normal patterns in a dataset. These deviations often point to critical incidents such as financial fraud, system failures, cyber intrusions, or health anomalies. [1, 2]
------------------------------
## Types of Anomalies and Examples## 1. Point Anomalies (Global Outliers)

* Definition: A single data instance that falls very far outside the distribution of the rest of the dataset.
* Example: Receiving a single credit card transaction for $10,000 when your typical monthly bill is around $200, or a human height reading of 8 feet. [3, 4, 5, 6]

## 2. Contextual Anomalies (Conditional Outliers)

* Definition: A data point that is considered normal in one context (e.g., time or location) but becomes abnormal in a different context.
* Example: A temperature reading of 30°C is completely normal during the summer, but would be a contextual anomaly if it occurs in the middle of winter. Similarly, a large server data transfer is standard during backup hours at midnight, but abnormal at 3:00 PM during peak business hours. [4, 6, 7]

## 3. Collective Anomalies

* Definition: A collection of related data points that do not look anomalous in isolation, but together form an abnormal pattern.
* Example: A single small file download or an isolated login attempt might look totally routine on its own. However, a rapid, coordinated sequence of hundreds of small file downloads together representing an exfiltration of a full customer database is a collective anomaly. [4, 8]


'''
```

```python
'''
Question 2: Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of
their approach and suitable use cases.
'''
'''
Isolation Forest, DBSCAN, and Local Outlier Factor (LOF) are popular unsupervised algorithms for anomaly and outlier detection, each using a distinct core approach and suited for different data structures. [1, 2, 3]
------------------------------
## 1. Isolation Forest

* Approach (Partition-Based & Global):
* Isolates observations by recursively and randomly selecting a feature and a split value between its maximum and minimum values.
   * Anomalies sit far from normal points and require fewer random splits to become isolated (yielding shorter path lengths in the trees).
   * Does not rely on distance or density calculations. [1, 4, 5]
* Suitable Use Cases:
* High-dimensional datasets (scales well with $O(n)$ time complexity).
   * Large-scale global anomaly detection (e.g., credit card fraud, network intrusion, sensor monitoring). [1, 2, 5]

------------------------------
## 2. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

* Approach (Density- & Cluster-Based):
* Groups together points that are closely packed (high-density regions) based on two parameters: $\epsilon$ (epsilon, the neighborhood radius) and min_samples (the minimum points required to form a dense region).
   * Points that do not belong to any dense cluster (falling into sparse regions or below the threshold) are automatically classified as noise/anomalies. [1, 2]
* Suitable Use Cases:
* Spatial or geographic data (e.g., GPS tracking, spatial clustering).
   * Datasets with arbitrary, non-linear cluster shapes where background noise needs to be separated. [1, 2]

------------------------------
## 3. Local Outlier Factor (LOF)

* Approach (Local Density-Based):
* Compares the local density of a given data point to the local densities of its $k$-nearest neighbors.
   * If a point has a significantly lower density than its immediate neighbors, it receives a high LOF score and is flagged as a local anomaly (even if that region's overall density differs from other clusters in the dataset). [1, 5, 6, 7]
* Suitable Use Cases:
* Datasets with varying densities across different clusters.
   * Finding subtle local outliers that global algorithms (like Isolation Forest) might miss because the point is not extreme relative to the entire dataset. [1, 2, 3, 5, 7]

------------------------------
## Summary Comparison

| Algorithm | Core Mechanism | Outlier Type Target | Dimensionality / Scaling |

| Isolation Forest | Random feature partitioning / tree path length | Global outliers | High-dimensional, fast $O(n)$ |
| DBSCAN | Density reachability ($\epsilon$, min_samples) | Noise / sparse points | Spatial/geometric data; sensitive to scaling |
| Local Outlier Factor | $k$-NN local density comparison | Local anomalies | Computationally heavier ($O(n^2)$); varies by density |

If you're working on a specific dataset, let me know:

* What kind of features or data you have (e.g., tabular, spatial, high-dimensional)
* Whether you expect global or local anomalies



'''
```

```python
'''
Question 3: What are the key components of a Time Series? Explain each with one
example.

'''
'''


[What is Time Series and Components of Time Series | by Ritu ...](https://medium.com/@ritusantra/what-is-time-series-and-components-of-time-series-c80b69ad5cb9)
[Time Series - Definition, Analysis, Forecasting, Components](https://www.wallstreetmojo.com/time-series/)
[Time Series Analysis: Definition, Components and Examples](https://www.analyticsvidhya.com/blog/2021/10/a-comprehensive-guide-to-time-series-analysis/)
[Time Series Analysis and Forecasting - GeeksforGeeks](https://www.geeksforgeeks.org/machine-learning/time-series-analysis-and-forecasting/)
[Unleash Components of Time Series - Key Secrets 2020](https://itfeature.com/time-series/introduction/components-of-time-series/)

A time series breaks down into four key components: Trend, Seasonality (Seasonal Variation), Cyclical Variation, and Irregular (Random/Noise) Variation. [1, 2]
------------------------------
## 1. Trend (Secular Trend)

* What it is: The long-term upward, downward, or flat movement in a dataset over an extended period (usually years).
* Example: A steady, multi-year increase in global temperatures due to climate change, or a gradual rise in e-commerce sales over a decade. [2, 3, 4]

## 2. Seasonality (Seasonal Variation)

* What it is: Short-term, repetitive patterns that occur at fixed, regular intervals (such as daily, weekly, monthly, or yearly) driven by weather, calendar dates, or holidays.
* Example: A spike in retail sales every December during the holiday shopping season, or higher sales of air conditioners during the summer months. [2, 4, 5]

## 3. Cyclical Variation (Cyclical Component)

* What it is: Long-term, wave-like fluctuations around a trend line that recur over irregular intervals longer than a year, typically tied to economic or business cycles (prosperity, recession, depression, recovery).
* Example: Multi-year housing market booms and busts or national economic recessions happening every 7 to 10 years. [2, 4, 6]

## 4. Irregular Variation (Random or Noise Component)

* What it is: Unpredictable, random, and short-term fluctuations left over after removing trend, seasonal, and cyclical influences.
* Example: A sudden drop in a local restaurant's daily revenue caused by an unexpected flash flood or a surprise public transit strike. [2, 4, 5, 6]


'''
```

```python
'''
Question 4: Define Stationary in time series. How can you test and transform a
non-stationary series into a stationary one?
'''
'''
A stationary time series is one whose statistical properties—such as mean, variance, and autocovariance—remain constant over time, meaning the process does not depend on when it is observed. Non-stationary series, by contrast, exhibit trends, drifting means, changing variance, or seasonality. [1, 2, 3]
------------------------------
## How to Test for Stationarity
You can detect non-stationarity using visual tools and formal statistical tests: [1]

* Visual Inspection: Plot the time series to check for obvious upward or downward trends, shifting variance, or repeating seasonal patterns. Examine the Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots. [1, 4]
* Augmented Dickey-Fuller (ADF) Test:
* Null Hypothesis ($H_0$): The series has a unit root (it is non-stationary).
   * Decision: If the $p$-value is $\le 0.05$ (and the ADF test statistic is less than the critical value), you reject $H_0$ and conclude the series is stationary. [3, 4, 5]
* Kwiatkowski-Phillips-Schmidt-Shin (KPSS) Test:
* Null Hypothesis ($H_0$): The series is stationary around a level or trend.
   * Decision: If the $p$-value is $> 0.05$, you fail to reject $H_0$, indicating stationarity. (Note: It reverses the null hypothesis compared to the ADF test.) [2, 4, 6]

------------------------------
## How to Transform a Non-Stationary Series into a Stationary One
When data is non-stationary, apply the following transformations: [2, 7]

   1. Variance Stabilization (Log / Power Transformations):
   * Use logarithmic, square root, or cube root transformations if the variance grows exponentially or fluctuates with the magnitude of the data.
      * Tip: Always apply log or power transformations before differencing. [8, 9]
   2. Differencing:
   * Subtract the previous observation from the current observation ($y_t - y_{t-1}$) to remove linear trends.
      * If a trend persists, apply second-order differencing (the difference of the differences).
      * Use seasonal differencing (subtracting the value from one seasonal cycle ago, e.g., $y_t - y_{t-s}$) to eliminate seasonal patterns. [3, 5, 8, 10, 11]
   3. Detrending / Decomposing:
   * Fit and subtract a deterministic trend line or use time series decomposition (trend, seasonal, and residual components) to isolate the stationary residual component. [12]



'''
```

```python
'''
Question 5: Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in
terms of structure and application.

'''
'''
AR, MA, ARIMA, SARIMA, and SARIMAX are a progressive family of time-series forecasting models that build upon one another by adding components for stationarity, seasonality, and external factors. [1, 2, 3]
------------------------------
## Model Structures

* AR (Autoregressive - $p$):
* Structure: Forecasts the future value based on a linear combination of its own past (lagged) values. The parameter $p$ defines how many past time steps are included. [2, 4]
* MA (Moving Average - $q$):
* Structure: Forecasts the future value using past forecast errors (white noise shocks) rather than past raw values. The parameter $q$ defines the number of lagged error terms. [2, 5]
* ARIMA (Autoregressive Integrated Moving Average - $p, d, q$):
* Structure: Combines AR ($p$) and MA ($q$) with an Integrated ($d$) differencing step. Differencing ($d$) subtracts past values to transform non-stationary data (data with a trending mean or variance) into stationary data. [1, 2, 3, 6]
* SARIMA (Seasonal ARIMA - $p, d, q$ & $P, D, Q, s$):
* Structure: Extends ARIMA by adding seasonal components. It includes seasonal autoregression ($P$), seasonal differencing ($D$), seasonal moving average ($Q$), and the seasonal cycle/frequency length ($s$). [1, 2, 6, 7]
* SARIMAX (SARIMA with Exogenous Variables):
* Structure: Extends SARIMA by adding exogenous regressors ($X$). These are external, independent variables (such as holiday markers, promotions, or temperature) that influence the target time series. [1, 3, 7]

------------------------------
## Applications and Use Cases

| Model | Core Use Case | Best Suited For |

| AR | Modeling momentum or inertia | Stationary data where current values depend strongly on immediate predecessors. |
| MA | Smoothing out random shocks or noise | Stationary data affected by short-term, unexpected random shocks. |
| ARIMA | Trend-based forecasting without seasonality | Stock prices (short-term), economic indicators, or non-seasonal data with a clear upward/downward trend. |
| SARIMA | Cyclic and seasonal pattern forecasting | Monthly airline passenger counts, retail sales with annual holidays, or quarterly energy usage. |
| SARIMAX | Complex multi-variable seasonal forecasting | Electricity grid demand (accounting for temperature/holidays) or retail sales (accounting for ad spend and weather). |


'''
```

```python
'''
Question 6: Load a time series dataset (e.g., AirPassengers), plot the original series,
and decompose it into trend, seasonality, and residual components
(Include your Python code and output in the code box below.)
'''
import matplotlib.pyplot as plt
import pandas as pd
import statsmodels.api as sm
from statsmodels.tsa.seasonal import seasonal_decompose

# Load the classic AirPassengers dataset via a direct URL
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url, index_col="Month", parse_dates=True)
df.rename(columns={"Passengers": "Passengers"}, inplace=True)
df.index.freq = "MS"  # Month Start frequency

# Decompose the time series (multiplicative model handles increasing amplitude)
decomposition = seasonal_decompose(df["Passengers"], model="multiplicative")

# Plot the original series and components
fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

decomposition.observed.plot(ax=axes[0], title="Observed (Original Series)")
decomposition.trend.plot(ax=axes[1], title="Trend Component")
decomposition.seasonal.plot(ax=axes[2], title="Seasonal Component")
decomposition.resid.plot(ax=axes[3], title="Residual Component")

plt.tight_layout()
plt.show()
```

```python
'''
Question 7: Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
detect anomalies. Visualize the anomalies on a 2D scatter plot.
(Include your Python code and output in the code box below.)
'''

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# 1. Generate synthetic NYC Taxi Fare dataset (distance vs. fare)
np.random.seed(42)
n_samples = 300

# Generate normal trips (typical linear relationship with noise)
distance_normal = np.random.uniform(0.5, 10, n_samples)
fare_normal = 2.5 + 3.0 * distance_normal + np.random.normal(0, 2, n_samples)

# Generate intentional anomalies
# Group A: Short distance but excessively high fare (e.g., predatory pricing or severe traffic)
distance_anom1 = np.random.uniform(0.5, 2, 10)
fare_anom1 = np.random.uniform(40, 60, 10)

# Group B: Long distance but impossibly low fare (e.g., meter tracking failure)
distance_anom2 = np.random.uniform(8, 12, 10)
fare_anom2 = np.random.uniform(5, 10, 10)

# Combine into a single structured DataFrame
distances = np.concatenate([distance_normal, distance_anom1, distance_anom2])
fares = np.concatenate([fare_normal, fare_anom1, fare_anom2])
df = pd.DataFrame({'trip_distance': distances, 'fare_amount': fares})

# 2. Configure and Fit the Isolation Forest Model
X = df[['trip_distance', 'fare_amount']].values
# contamination represents the estimated proportion of outliers in the dataset
iso_forest = IsolationForest(contamination=0.06, random_state=42)
df['anomaly_label'] = iso_forest.fit_predict(X)

# 3. Separate the dataset into normal points and flagged anomalies
normal_data = df[df['anomaly_label'] == 1]
anomalies = df[df['anomaly_label'] == -1]

# 4. Visualize the results on a 2D Scatter Plot
plt.figure(figsize=(10, 6))
plt.scatter(normal_data['trip_distance'], normal_data['fare_amount'],
            c='lightgray', label='Normal Trips', alpha=0.8, edgecolors='k', s=40)
plt.scatter(anomalies['trip_distance'], anomalies['fare_amount'],
            c='red', label='Detected Anomalies', alpha=0.9, edgecolors='k', s=60, marker='X')

plt.title('NYC Taxi Fare Anomaly Detection using Isolation Forest', fontsize=14, fontweight='bold')
plt.xlabel('Trip Distance (Miles)', fontsize=12)
plt.ylabel('Fare Amount ($)', fontsize=12)
plt.legend(loc='upper left', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Print class distribution summaries
print("Prediction Summary Count:")
print(df['anomaly_label'].value_counts().rename({1: 'Normal Data (1)', -1: 'Anomaly (-1)'}))
```

```python
'''
Question 8: Train a SARIMA model on the monthly airline passengers dataset.
Forecast the next 12 months and visualize the results.
(Include your Python code and output in the code box below.)
'''
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# 1. Load the Monthly Airline Passengers dataset via a direct URL
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url, parse_dates=['Month'], index_col='Month')
df.index.freq = 'MS'  # Set the monthly start frequency

# 2. Define and fit the SARIMA model
# A standard specification for this dataset is SARIMA(1,1,1)x(1,1,1,12)
model = sm.tsa.statespace.SARIMAX(df['Passengers'],
                                  order=(1, 1, 1),
                                  seasonal_order=(1, 1, 1, 12),
                                  enforce_stationarity=False,
                                  enforce_invertibility=False)
results = model.fit(disp=False)

# 3. Forecast the next 12 months
forecast_steps = 12
forecast = results.get_forecast(steps=forecast_steps)
forecast_mean = forecast.predicted_mean
confidence_intervals = forecast.conf_int()

# Print the forecast output
print("--- Forecasted Values (Next 12 Months) ---")
print(forecast_mean)

# 4. Visualize the results
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['Passengers'], label='Historical Data', color='blue')
plt.plot(forecast_mean.index, forecast_mean, label='SARIMA Forecast', color='red', linestyle='--')
plt.fill_between(confidence_intervals.index,
                 confidence_intervals.iloc[:, 0],
                 confidence_intervals.iloc[:, 1],
                 color='pink', alpha=0.3, label='95% Confidence Interval')

plt.title('Monthly Airline Passengers - 12-Month SARIMA Forecast')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

```python
'''
Question 9: Apply Local Outlier Factor (LOF) on any numerical dataset to detect
anomalies and visualize them using matplotlib.
(Include your Python code and output in the code box below.)
'''


import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import LocalOutlierFactor

# 1. Generate a synthetic numerical dataset with normal clusters and outliers
np.random.seed(42)
X_inliers = 0.3 * np.random.randn(100, 2)
X_outliers = np.random.uniform(low=-4, high=4, size=(20, 2))
X = np.r_[X_inliers + 2, X_inliers - 2, X_outliers]

# 2. Apply Local Outlier Factor (LOF)
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)
y_pred = lof.fit_predict(X)  # -1 for outliers, 1 for inliers
X_scores = lof.negative_outlier_factor_  # Higher negative score means more normal; lower negative score means outlier

# 3. Visualize results with Matplotlib
plt.figure(figsize=(8, 6))

# Plot normal points
plt.scatter(
    X[y_pred == 1, 0], X[y_pred == 1, 1], color='blue', s=30, label='Inliers'
)

# Plot anomalies / outliers
plt.scatter(
    X[y_pred == -1, 0],
    X[y_pred == -1, 1],
    color='red',
    s=50,
    edgecolor='k',
    marker='x',
    label='Outliers',
)

# Circle size proportional to outlier score
radius = (X_scores.max() - X_scores) / (X_scores.max() - X_scores.min())
plt.scatter(
    X[:, 0],
    X[:, 1],
    s=1000 * radius,
    edgecolors='r',
    facecolors='none',
    label='Outlier Score Magnitude',
)

plt.title('Local Outlier Factor (LOF) Anomaly Detection')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
'''
Question 10: You are working as a data scientist for a power grid monitoring company.
Your goal is to forecast energy demand and also detect abnormal spikes or drops in
real-time consumption data collected every 15 minutes. The dataset includes features
like timestamp, region, weather conditions, and energy usage.
Explain your real-time data science workflow:
● How would you detect anomalies in this streaming data (Isolation Forest / LOF /
DBSCAN)?
● Which time series model would you use for short-term forecasting (ARIMA /
SARIMA / SARIMAX)?
● How would you validate and monitor the performance over time?
● How would this solution help business decisions or operations?
'''
'''
Here is the end-to-end data science workflow designed to handle 15-minute streaming power grid data, combining real-time anomaly detection, short-term forecasting, robust validation, and strategic business applications.
------------------------------
## 1. Real-Time Anomaly Detection
For 15-minute streaming energy data, Isolation Forest is the most viable option among the three choices.

* Why Isolation Forest wins: Power consumption data changes rapidly. Isolation Forest isolates anomalies by randomly partitioning feature spaces using tree structures. Since anomalies require fewer splits to isolate, they appear closer to the root of the trees. It handles high-dimensional streaming data efficiently, scales well linearly, and can be adapted to streaming data via sliding-window retraining.
* Why the others fall short for streaming:
* Local Outlier Factor (LOF) and DBSCAN are distance- and density-based algorithms.
   * They require calculating distances between all data points ($O(N^2)$ complexity), making them computationally too expensive to run sequentially on fast-moving streams.
* Implementation Strategy: Compute rolling baseline features (e.g., the last 2 hours of consumption vs. historical averages for that specific day/hour) and feed them into the Isolation Forest to instantly flag sudden spikes or drops.

------------------------------
## 2. Short-Term Time Series Forecasting
For this application, SARIMAX (Seasonal AutoRegressive Integrated Moving Average with eXogenous regressors) is the ideal model.

* Why SARIMAX is the best choice: Energy consumption is highly dependent on both seasonality (time of day, day of the week) and external variables (weather conditions like temperature, humidity, and solar radiation).
* Model breakdown:
* SARIMA captures the daily (96 periods per day for 15-minute intervals) and weekly cycles.
   * The X factor (Exogenous variables) allows the model to inject weather data. For example, a sudden heatwave will immediately spike air conditioning demand; SARIMAX can preemptively account for this, whereas ARIMA or standard SARIMA would lag behind until the spike actually occurred.
* Production Note: Because calculating 96-period seasonal cycles can be computationally heavy in standard SARIMAX, in a modern enterprise grid, this might be supplemented or replaced by Prophet or LightGBM/XGBoost with lagging features for faster real-time inference.

------------------------------
## 3. Validation and Monitoring Strategy
A streaming pipeline requires continuous validation to combat data drift (changing weather patterns) and concept drift (changing consumer habits).

* Validation Protocol: Use Time Series Forward Chaining Cross-Validation (Rolling Window Validation). Never use standard K-Fold cross-validation, as shuffling the data breaks the temporal dependency and causes data leakage. Train on a rolling past window (e.g., past 30 days) and test on the next horizon (e.g., next 24 hours).
* Performance Monitoring:
* Track accuracy metrics like MAPE (Mean Absolute Percentage Error) and RMSE (Root Mean Squared Error) on a live dashboard.
   * Set up automated triggers to retrain the SARIMAX model if MAPE exceeds a pre-defined threshold (e.g., >5%) over a sustained 6-hour period.
* Anomaly Monitoring: Monitor the percentage of data points flagged as anomalies. If the anomaly rate suddenly jumps from 1% to 15%, it flags a systemic issue (e.g., a broken grid sensor or a major blackout) rather than a normal outlier.

------------------------------
## 4. Business Decisions and Operations Impact
This dual-purpose workflow directly translates data science into financial savings and grid reliability:

* Preventing Blackouts & Load Balancing: Accurate short-term forecasting allows grid operators to spin up auxiliary power plants or buy energy ahead of time if a massive surge is predicted.
* Infrastructure Asset Protection: Real-time anomaly detection identifies sudden, dangerous spikes that indicate equipment failure, short circuits, or transformer overloads, allowing automated systems to reroute power before physical damage occurs.
* Cost Optimization: Buying energy on the spot market last-minute is incredibly expensive. Preemptive forecasting allows utility companies to optimize their energy procurement strategies, saving millions annually.
* Theft and Fraud Detection: Persistent, unexplained drops in localized nodes can instantly alert field operations to power theft or malfunctioning smart meters.
'''
```
